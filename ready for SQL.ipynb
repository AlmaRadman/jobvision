```python
import pyodbc

server = 'localhost'   
database = 'master'   

try:
    conn = pyodbc.connect(
        f'DRIVER={{ODBC Driver 18 for SQL Server}};SERVER={server};DATABASE={database};Trusted_Connection=yes;'
    )
    cursor = conn.cursor()
    print("✅ Connected to SQL Server!")
except Exception as e:
    print("❌ Connection failed:", e)

```
